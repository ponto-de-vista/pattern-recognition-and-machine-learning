```python
query = rf"""
    WITH ibge_lookup AS (
        SELECT 
            CAST(cod_ibge AS INT) AS cod_ibge,
            municipio AS nome_original_ibge,
            -- Vacina do IBGE: Limpa acentos, hífens e apóstrofos
            UPPER(TRIM(regexp_replace(strip_accents(REPLACE(REPLACE(municipio, '-', ' '), '''', ' ')), '\s+', ' ', 'g'))) AS municipio_norm
        FROM read_csv(
            'codigos_municipios_regioes.csv', -- Usando o nome correto do seu arquivo
            header=True, 
            encoding='iso-8859_9-1999', 
            ignore_errors=True
        )
        WHERE length(CAST(cod_ibge AS VARCHAR)) > 4
    ),
    despesas_municipios AS (
        SELECT DISTINCT ds_municipio
        FROM read_parquet('./despesas-datasets/despesas-*.parquet')
        WHERE ds_municipio IS NOT NULL
    )
    SELECT 
        d.ds_municipio AS cidade_na_despesa
    FROM despesas_municipios d
    LEFT JOIN ibge_lookup ibge 
        -- Vacina das Despesas: Corrige Florínea, São Luiz e limpa pontuações
        ON UPPER(TRIM(regexp_replace(strip_accents(
            REPLACE(
                REPLACE(
                    REPLACE(
                        REPLACE(d.ds_municipio, 'Florínia', 'Florínea'), 
                    'São Luís do', 'São Luiz do'),
                '-', ' '),
            '''', ' ')
        ), '\s+', ' ', 'g'))) = ibge.municipio_norm
    WHERE ibge.cod_ibge IS NULL 
    ORDER BY d.ds_municipio;
"""

df_erros = duckdb.sql(query).df()

if df_erros.empty:
    print("✅ PERFEITO! O código ajustou todas as inconsistências.")
    print("Nenhuma cidade das Despesas ficou sem o código IBGE.")
else:
    print(f"⚠️ Atenção! Encontradas {len(df_erros)} cidades que ainda não bateram:")
    print(df_erros.to_string(index=False))
```

```python
print("Procurando cidades órfãs nas despesas...")

query = rf"""
    WITH ibge_lookup AS (
        SELECT 
            CAST(cod_ibge AS INT) AS cod_ibge,
            UPPER(TRIM(regexp_replace(strip_accents(REPLACE(municipio, '-', ' ')), '\s+', ' ', 'g'))) AS municipio_norm
        FROM read_csv(
            'codigos_municipios_regioes.csv', 
            header=True, 
            encoding='iso-8859_9-1999', 
            ignore_errors=True
        )
        WHERE length(CAST(cod_ibge AS VARCHAR)) > 4
    ),
    despesas_municipios AS (
        SELECT DISTINCT ds_municipio
        FROM read_parquet('./despesas-datasets/despesas-*.parquet')
        WHERE ds_municipio IS NOT NULL
    )
    SELECT 
        d.ds_municipio AS cidade_na_despesa
    FROM despesas_municipios d
    LEFT JOIN ibge_lookup ibge 
        ON UPPER(TRIM(regexp_replace(strip_accents(REPLACE(d.ds_municipio, '-', ' ')), '\s+', ' ', 'g'))) = ibge.municipio_norm
    WHERE ibge.cod_ibge IS NULL 
    ORDER BY d.ds_municipio;
"""

df_erros = duckdb.sql(query).df()

if df_erros.empty:
    print("🎉 Sucesso absoluto! Todas as cidades bateram com o IBGE.")
else:
    print(f"⚠️ Encontradas {len(df_erros)} cidades sem match:")
    print(df_erros.to_string(index=False))
```

```python
print("📊 Iniciando Auditoria de Qualidade dos Parquets...\n")

# Cabeçalho da nossa tabela de resultados
print(f"{'Ano':<6} | {'Linhas Originais':<16} | {'Linhas Novo':<14} | {'Status Linhas':<14} | {'NULLs em cod_ibge'}")
print("-" * 75)

with duckdb.connect(":memory:") as con:
    for year in range(2009, 2019):
        orig = f'./despesas-datasets/despesas-{year}.parquet'
        novo = f'./despesas-datasets/despesas-{year}-ibge.parquet'

        if os.path.exists(orig) and os.path.exists(novo):
            # Conta as linhas do arquivo original
            qtd_orig = con.sql(f"SELECT count(*) FROM '{orig}'").fetchone()[0]
            
            # Conta as linhas do arquivo novo
            qtd_novo = con.sql(f"SELECT count(*) FROM '{novo}'").fetchone()[0]
            
            # Conta quantos cod_ibge ficaram vazios
            nulls = con.sql(f"SELECT count(*) FROM '{novo}' WHERE cod_ibge IS NULL").fetchone()[0]
            
            # Validações
            status_linhas = "✅ Bateu" if qtd_orig == qtd_novo else f"❌ Diff: {qtd_orig - qtd_novo}"
            status_nulls = "✅ Zero" if nulls == 0 else f"❌ {nulls} Nulls"
            
            print(f"{year:<6} | {qtd_orig:<16} | {qtd_novo:<14} | {status_linhas:<14} | {status_nulls}")
        else:
            print(f"{year:<6} | ⚠️ Arquivos não encontrados para comparação.")
```

```python
import duckdb
import glob

print("🔍 Verificando consistência de colunas entre todos os Parquets...\n")

arquivos = sorted(glob.glob('./despesas-datasets/despesas-*.parquet'))
esquemas = {}

with duckdb.connect(":memory:") as con:
    for arquivo in arquivos:
        ano = arquivo.split('-')[1]
        # Pegamos o nome da coluna e o tipo para garantir 100% de match
        df_schema = con.sql(f"DESCRIBE SELECT * FROM '{arquivo}'").df()
        
        # Criamos uma assinatura do arquivo (coluna: tipo)
        assinatura = list(zip(df_schema['column_name'], df_schema['column_type']))
        esquemas[ano] = assinatura

# Comparação
primeiro_ano = list(esquemas.keys())[0]
esquema_referencia = esquemas[primeiro_ano]
todas_iguais = True

print(f"{'Ano':<6} | {'Qtd Colunas':<12} | {'Status'}")
print("-" * 35)

for ano, esquema in esquemas.items():
    if esquema == esquema_referencia:
        status = "✅ Consistente"
    else:
        status = "❌ DIVERGENTE"
        todas_iguais = False
    print(f"{ano:<6} | {len(esquema):<12} | {status}")

if todas_iguais:
    print(f"\n🚀 Sucesso! Todos os {len(arquivos)} arquivos possuem as mesmas colunas e tipos.")
    print("\nEstrutura de referência (usada no merge):")
    for col, tipo in esquema_referencia:
        print(f"- {col}: {tipo}")
else:
    print("\n⚠️ Alerta: Existem diferenças entre os anos. Precisaremos normalizar as colunas antes do merge.")
```

```python
import duckdb

parquet_file = './despesas-datasets/despesas-2018.parquet'

print(f"🕵️ Análise comparativa de integridade: {parquet_file}\n")

with duckdb.connect(":memory:") as con:
    # 1. Verificar se existem valores diferentes (ignorando zeros ou nulos se houver)
    query_conflito = f"""
        SELECT 
            ds_municipio, 
            cod_ibge, 
            codigo_municipio_ibge,
            COUNT(*) as ocorrencias
        FROM '{parquet_file}'
        WHERE CAST(cod_ibge AS VARCHAR) != CAST(codigo_municipio_ibge AS VARCHAR)
          AND codigo_municipio_ibge IS NOT NULL
        GROUP BY 1, 2, 3
        LIMIT 20;
    """
    
    # 2. Verificar a taxa de preenchimento da coluna intrusa
    query_stats = f"""
        SELECT 
            COUNT(*) as total_linhas,
            COUNT(codigo_municipio_ibge) as preenchidos,
            COUNT(DISTINCT codigo_municipio_ibge) as codigos_unicos_intrusa,
            COUNT(DISTINCT cod_ibge) as codigos_unicos_auditados
        FROM '{parquet_file}';
    """

    print("📊 Estatísticas Gerais:")
    stats = con.sql(query_stats).df()
    print(stats.to_string(index=False))

    print("\n🔍 Detectando Divergências (Matches Errados):")
    conflitos = con.sql(query_conflito).df()
    
    if conflitos.empty:
        print("✅ SUCESSO: O 'codigo_municipio_ibge' condiz 100% com o seu 'cod_ibge' auditado (onde está preenchido).")
    else:
        print("⚠️ ATENÇÃO: Encontradas divergências entre as colunas!")
        print(conflitos.to_string(index=False))
```

```python
import duckdb
import pandas as pd

def analisar_evolucao_esquema(diretorio):
    with duckdb.connect(":memory:") as con:
        # Pega a lista de colunas de cada ano
        anos = range(2009, 2019)
        mapa_colunas = {}
        
        for ano in anos:
            path = f"{diretorio}/despesas-{ano}.parquet"
            # DESCRIBE é mais rápido que SELECT * LIMIT 0
            cols = con.sql(f"DESCRIBE SELECT * FROM '{path}'").df()['column_name'].tolist()
            mapa_colunas[str(ano)] = cols

        # Criar DataFrame de presença (True/False)
        todas_colunas = sorted(list(set([col for sublist in mapa_colunas.values() for col in sublist])))
        df_comp = pd.DataFrame(index=todas_colunas)
        
        for ano, cols in mapa_colunas.items():
            df_comp[ano] = df_comp.index.isin(cols)
        
        # Filtrar apenas colunas que NÃO estão em todos os anos (as divergentes)
        divergentes = df_comp[df_comp.sum(axis=1) != len(anos)]
        
        print("📊 COLUNAS DIVERGENTES ENCONTRADAS:")
        return divergentes

# Uso:
df_analise = analisar_evolucao_esquema('./despesas-datasets')
display(df_analise)
```
